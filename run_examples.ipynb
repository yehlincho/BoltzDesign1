## BoltzDesign — run examples (boltz1 & boltz2)

Design recipe (Gumbel init, learning rate, e_soft, helix range) is **auto-applied from the per-target-type config** (`boltzdesign/configs/`) — you only specify the target + model.

- **Models:** `--boltz_model_version boltz2` (default) or `boltz1`.
- **Templates:** boltz2 only (boltz1 does not support templates).
- **Default helix_loss:** (-0.3, 0.0); override with `--helix_loss_min/max`.
